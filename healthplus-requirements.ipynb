
## Phase 4

Install dependencies in your terminal after activating your virtual environment
```bash
pip install pandas numpy sqlalchemy psycopg2-binary ipykernel
pip freeze > requirements.txt 
```

- The second command (pip freeze > requirements.txt): creates a record of all the Python packages and their versions used in the project, making it easy for anyone to recreate the same environment and run the project successfully.
